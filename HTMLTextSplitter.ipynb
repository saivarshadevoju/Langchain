# How to Split by HTML Header

HTMLHeaderTextSplitter is a "structure-aware" chunker that splits text at the HTML element level and adds metadata for each header "relevant" to any given chunk. It can return chunks element by element or combine elements with the same metadata, with the objectives of (a) keeping related text grouped (more or less) semantically and (b) preserving context-rich information encoded in document structures. It can be used with other text splitters as part of a chunking pipeline.

```python
from langchain_text_splitters import HTMLHeaderTextSplitter
html_string = """
<!DOCTYPE html>
<html>
<head>
    <title>AI Learning Page</title>
</head>
<body>

    <h1>Artificial Intelligence</h1>

    <div>
        <p>
            Artificial Intelligence (AI) is the simulation of human intelligence
            in machines that are programmed to think and learn.
        </p>
    </div>

    <h2>Types of AI</h2>

    <div>
        <p>There are mainly three types of AI:</p>

        <ul>
            <li>Narrow AI</li>
            <li>General AI</li>
            <li>Super AI</li>
        </ul>
    </div>

    <h2>Applications of AI</h2>

    <div>
        <p>AI is widely used in different industries.</p>

        <ol>
            <li>Healthcare</li>
            <li>Education</li>
            <li>Finance</li>
            <li>Agriculture</li>
        </ol>
    </div>

    <h3>AI in Healthcare</h3>

    <div>
        <p>
            AI helps doctors in disease prediction, medical imaging,
            and patient monitoring systems.
        </p>
    </div>

    <h3>AI in Education</h3>

    <div>
        <p>
            AI-powered platforms provide personalized learning experiences
            for students.
        </p>
    </div>

    <h2>Advantages of AI</h2>

    <div>
        <ul>
            <li>Automation of repetitive tasks</li>
            <li>Faster decision making</li>
            <li>Reduced human errors</li>
            <li>24/7 availability</li>
        </ul>
    </div>

    <h2>Challenges of AI</h2>

    <div>
        <ul>
            <li>Job displacement</li>
            <li>Privacy concerns</li>
            <li>Bias in AI models</li>
            <li>High development cost</li>
        </ul>
    </div>

</body>
</html>
"""
headers_to_split = [
    ("h1", "Artificial Intelligence"),
    ("h2", "Types of AI")
]
splitter = HTMLHeaderTextSplitter(headers_to_split)
docs = splitter.split_text(html_string)
docs
```

```python
import requests

url = "https://en.wikipedia.org/wiki/Main_Page" 
headers_to_split = [
    ("h1", "Artificial Intelligence"),
    ("h2", "Types of AI")
]
splitter = HTMLHeaderTextSplitter(headers_to_split)
response = requests.get(url, headers={"User-Agent": "Mozilla/5.0"})
docs = splitter.split_text(response.text)
docs
```
