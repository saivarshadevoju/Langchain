# Ollama

Ollama supports embedding models, making it possible to build RAG applications that combine text prompts with existing documents or other data.

```python
from langchain_community.embeddings import OllamaEmbeddings
embeddings = (
    OllamaEmbeddings(model = 'tinyllama') ## by default it uses llama2
) 
```

```python
embeddings
```

```python
r1 = embeddings.embed_documents([
    "What is AI",
    "What is machine learning",
])
```

```python
r1
```

```python
len(r1[0])
```

```python
embeddings.embed_query("What is AI")
```
